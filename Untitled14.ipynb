```python
# =========================================================
# HIERARCHICAL CLUSTERING USING GOWER DISTANCE
# FOR MIXED NUMERICAL + CATEGORICAL DATA
# =========================================================

!pip install gower

import pandas as pd
import numpy as np
import gower
import matplotlib.pyplot as plt
import seaborn as sns

from scipy.cluster.hierarchy import linkage, dendrogram, fcluster
from scipy.spatial.distance import squareform
from sklearn.metrics import silhouette_score



import warnings
warnings.filterwarnings('ignore')

# =========================================================
# LOAD DATA
# =========================================================

df = pd.read_csv('insurance2.csv')
# Change sex and smoker to True/False
df['sex'] = df['sex'].astype(bool)
df['smoker'] = df['smoker'].astype(bool)
# Returns a new DataFrame with duplicates removed
df = df.drop_duplicates()

# Convert 'insuranceclaim' to string labels for consistency
df['insuranceclaim'] = df['insuranceclaim'].replace({
    0: 'No Claim',
    1: 'Claim'
})

# =========================================================
# DEFINE VARIABLES
# =========================================================

# Target variable
target = 'insuranceclaim'

# Numerical variables
# (Recommended: exclude charges from clustering)
numerical_cols = ['age', 'bmi', 'children','charges']


# Categorical variables
categorical_cols = ['sex', 'region','smoker']


# =========================================================
# PREPARE DATA
# =========================================================

# Create feature matrix
X = df[numerical_cols + categorical_cols].copy()

# Convert categorical variables to string
# This ensures Gower treats them as categorical

for col in categorical_cols:
    X[col] = X[col].astype(str)

print("Feature Matrix Shape:", X.shape)

# =========================================================
# SPECIFY CATEGORICAL FEATURES
# =========================================================

# False = numerical
# True  = categorical

cat_features = [
    False,
     False, # age
    False,  # bmi
    False,  # children
    True,   # sex
      True, # smoker
    True    # region
]

# =========================================================
# COMPUTE GOWER DISTANCE MATRIX
# =========================================================

print("\nComputing Gower Distance Matrix...")

gower_dist = gower.gower_matrix(
    X,
    cat_features=cat_features
)

print("Gower Distance Matrix Shape:", gower_dist.shape)

print("Distance Range:",
      round(gower_dist.min(), 4),
      "to",
      round(gower_dist.max(), 4))

# =========================================================
# CONVERT TO CONDENSED DISTANCE MATRIX
# =========================================================

# linkage() requires condensed form

condensed_dist = squareform(
    gower_dist,
    checks=False
)

# =========================================================
# HIERARCHICAL CLUSTERING
# =========================================================

# Use average linkage
# (Correct for Gower distance)

linked = linkage(
    condensed_dist,
    method='average'
)

# =========================================================
# DENDROGRAM
# =========================================================

plt.figure(figsize=(14, 7))

dendrogram(
    linked,
    truncate_mode='level',
    p=5,
    color_threshold=0.3
)

plt.title(
    'Hierarchical Clustering Dendrogram\n(Gower Distance + Average Linkage)',
    fontsize=14
)

plt.xlabel('Observations')
plt.ylabel('Gower Distance')

plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()

# =========================================================
# SILHOUETTE ANALYSIS
# =========================================================

silhouette_scores = []

K_range = range(2, 11)

print("\nSilhouette Scores:\n")

for k in K_range:

    labels = fcluster(
        linked,
        k,
        criterion='maxclust'
    )

    score = silhouette_score(
        gower_dist,
        labels,
        metric='precomputed'
    )

    silhouette_scores.append(score)

    print(f"k = {k} --> Silhouette Score = {score:.4f}")

# =========================================================
# PLOT SILHOUETTE SCORES
# =========================================================

plt.figure(figsize=(8, 5))

plt.plot(
    K_range,
    silhouette_scores,
    marker='o',
    linewidth=2
)

plt.xlabel('Number of Clusters')
plt.ylabel('Silhouette Score')

plt.title(
    'Silhouette Analysis\nHierarchical Clustering with Gower Distance'
)

plt.xticks(K_range)

plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()

# =========================================================
# BEST NUMBER OF CLUSTERS
# =========================================================

best_k = K_range[np.argmax(silhouette_scores)]

print("\nBest Number of Clusters:", best_k)

# =========================================================
# ASSIGN FINAL CLUSTERS
# =========================================================

df['Cluster'] = fcluster(
    linked,
    best_k,
    criterion='maxclust'
) - 1

# =========================================================
# CLUSTER DISTRIBUTION
# =========================================================

print("\nCluster Distribution:\n")

print(df['Cluster'].value_counts())

# =========================================================
# NUMERICAL VARIABLE PROFILE
# =========================================================

print("\nCluster Means:\n")

print(
    df.groupby('Cluster')[
        ['age', 'bmi', 'children', 'charges']
    ].mean().round(2)
)

# =========================================================
# CATEGORICAL VARIABLE PROFILE
# =========================================================

print("\nSex Distribution (%):\n")

sex_profile = pd.crosstab(
    df['Cluster'],
    df['sex'],
    normalize='index'
) * 100

print(sex_profile.round(2))

# ---------------------------------------------------------

print("\nSmoker Distribution (%):\n")

smoker_profile = pd.crosstab(
    df['Cluster'],
    df['smoker'],
    normalize='index'
) * 100

print(smoker_profile.round(2))

# ---------------------------------------------------------

print("\nRegion Distribution (%):\n")

region_profile = pd.crosstab(
    df['Cluster'],
    df['region'],
    normalize='index'
) * 100

print(region_profile.round(2))

# ---------------------------------------------------------

print("\nInsurance Claim Distribution (%):\n")

claim_profile = pd.crosstab(
    df['Cluster'],
    df['insuranceclaim'],
    normalize='index'
) * 100

print(claim_profile.round(2))




```

## Machine Learning Model Building

Now we will prepare the data for building machine learning models for each cluster separately. This involves:

1.  Separating the original DataFrame into two based on the `Cluster` column.
2.  Defining features (X) and the target variable (y) for each cluster.
3.  Applying one-hot encoding to the categorical features to convert them into a numerical format suitable for most machine learning algorithms.

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

# Define numerical and categorical columns for ML models.
# 'charges' is now treated as a feature for predicting 'insuranceclaim'.
ml_numerical_cols = ['age', 'bmi', 'children', 'charges']
ml_categorical_cols = ['sex', 'region', 'smoker']
ml_target = 'insuranceclaim'

def preprocess_data(df_cluster, numerical_cols, categorical_cols, target_col):
    """Preprocesses data for a given cluster by performing one-hot encoding.

    Args:
        df_cluster (pd.DataFrame): DataFrame for a specific cluster.
        numerical_cols (list): List of numerical column names.
        categorical_cols (list): List of categorical column names.
        target_col (str): Name of the target column.

    Returns:
        tuple: X_processed (pd.DataFrame), y (pd.Series), preprocessor (ColumnTransformer)
    """

    # Define X and y
    X = df_cluster[numerical_cols + categorical_cols].copy()
    y = df_cluster[target_col].map({'No Claim': 0, 'Claim': 1})

    # Create a column transformer for one-hot encoding
    preprocessor = ColumnTransformer(
        transformers=[
            ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_cols)
        ],
        remainder='passthrough'  # Keep numerical columns as they are
    )

    # Apply preprocessing
    X_processed = preprocessor.fit_transform(X)

    # Get feature names after one-hot encoding
    cat_feature_names = preprocessor.named_transformers_['cat'].get_feature_names_out(categorical_cols)
    all_feature_names = list(cat_feature_names) + numerical_cols

    X_processed_df = pd.DataFrame(X_processed, columns=all_feature_names, index=X.index)

    return X_processed_df, y, preprocessor

# Separate data into clusters
df_cluster0 = df[df['Cluster'] == 0].copy()
df_cluster1 = df[df['Cluster'] == 1].copy()

print(f"Shape of Cluster 0 data: {df_cluster0.shape}")
print(f"Shape of Cluster 1 data: {df_cluster1.shape}")

# Preprocess Cluster 0 data
X0, y0, preprocessor0 = preprocess_data(df_cluster0, ml_numerical_cols, ml_categorical_cols, ml_target)
print(f"\nShape of preprocessed Cluster 0 features: {X0.shape}")
print("Cluster 0 target distribution:\n", y0.value_counts(normalize=True))

# Preprocess Cluster 1 data
X1, y1, preprocessor1 = preprocess_data(df_cluster1, ml_numerical_cols, ml_categorical_cols, ml_target)
print(f"\nShape of preprocessed Cluster 1 features: {X1.shape}")
print("Cluster 1 target distribution:\n", y1.value_counts(normalize=True))

# Display first few rows of preprocessed data for Cluster 0
print("\nPreprocessed Cluster 0 Features (first 5 rows):")
display(X0.head())

# Display first few rows of preprocessed data for Cluster 1
print("\nPreprocessed Cluster 1 Features (first 5 rows):")
display(X1.head())
```

## Handle Class Imbalance

As observed, Cluster 1 has a significant class imbalance (90.88% claims). To address this, we will apply SMOTE (Synthetic Minority Over-sampling Technique) to the training data of Cluster 1. Cluster 0 is relatively balanced, so no balancing technique will be applied.

```python
from imblearn.over_sampling import SMOTE
from collections import Counter
from sklearn.model_selection import StratifiedKFold, GridSearchCV
from sklearn.metrics import accuracy_score, precision_score, f1_score, recall_score, classification_report
from imblearn.pipeline import Pipeline # Corrected: Use Pipeline from imblearn.pipeline

# For Cluster 1, apply SMOTE to the training data during cross-validation.
# For now, we will just show the effect of SMOTE if applied to the full dataset.
# In the actual cross-validation loop, SMOTE will be applied only to training folds.

# Display original distribution for Cluster 1
print("Original Cluster 1 target distribution:", Counter(y1))

smote = SMOTE(random_state=42)
X1_resampled, y1_resampled = smote.fit_resample(X1, y1)

print("Resampled Cluster 1 target distribution (if applied to full data):", Counter(y1_resampled))

# We will use this information to decide whether to apply SMOTE or not in the CV function
# However, the actual SMOTE will be performed inside the cross-validation loop on training data only

# Define ML models to use
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.svm import SVC

models = {
    'Logistic Regression': {
        'model': LogisticRegression(random_state=42, solver='liblinear'),
        'params': {
            'C': [0.01, 0.1, 1, 10]
        }
    },
    'Decision Tree': {
        'model': DecisionTreeClassifier(random_state=42),
        'params': {
            'max_depth': [3, 5, 7, None]
        }
    },
    'Random Forest': {
        'model': RandomForestClassifier(random_state=42),
        'params': {
            'n_estimators': [50, 100, 200],
            'max_depth': [5, 10, None]
        }
    },
    'Gradient Boosting': {
        'model': GradientBoostingClassifier(random_state=42),
        'params': {
            'n_estimators': [50, 100, 200],
            'learning_rate': [0.01, 0.1, 0.2]
        }
    },
    'Support Vector Machine': {
        'model': SVC(random_state=42, probability=True),
        'params': {
            'C': [1],
            'kernel': ['rbf']
        }
    }
}

def train_evaluate_models(X_data, y_data, cluster_name, use_smote=False):
    """Trains and evaluates multiple ML models using K-fold cross-validation and GridSearchCV.

    Args:
        X_data (pd.DataFrame): Features for the cluster.
        y_data (pd.Series): Target variable for the cluster.
        cluster_name (str): Name of the cluster (e.g., 'Cluster 0', 'Cluster 1').
        use_smote (bool): Whether to apply SMOTE for balancing the training data.

    Returns:
        dict: A dictionary containing best models and their performance metrics.
    """
    print(f"\n--- Training and Evaluation for {cluster_name} ---")
    results = {}
    cv_strategy = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

    for model_name, model_info in models.items():
        print(f"\nBuilding {model_name}...")
        estimator = model_info['model']
        param_grid = model_info['params']

        # Create a pipeline if SMOTE is to be used
        if use_smote:
            pipeline_steps = [
                ('smote', SMOTE(random_state=42)),
                ('classifier', estimator)
            ]
            pipeline = Pipeline(pipeline_steps) # Now uses imblearn.Pipeline
            # Adjust parameter names for pipeline
            param_grid = {f'classifier__{key}': value for key, value in param_grid.items()}
            model_to_tune = pipeline
        else:
            model_to_tune = estimator

        grid_search = GridSearchCV(
            model_to_tune,
            param_grid,
            cv=cv_strategy,
            scoring='f1', # Focus on F1-score for evaluation
            n_jobs=-1, # Use all available cores
            verbose=1
        )

        grid_search.fit(X_data, y_data)

        best_model = grid_search.best_estimator_
        best_params = grid_search.best_params_
        print(f"Best Parameters for {model_name}: {best_params}")

        # Initialize lists to collect predictions for all folds
        y_pred_test_total = np.array([])
        y_true_test_total = np.array([])
        y_pred_train_total = np.array([])
        y_true_train_total = np.array([])

        # Collect predictions from each fold
        for train_index, test_index in cv_strategy.split(X_data, y_data):
            X_train, X_test = X_data.iloc[train_index], X_data.iloc[test_index]
            y_train, y_test = y_data.iloc[train_index], y_data.iloc[test_index]

            # Fit model on training data, applying SMOTE if necessary
            if use_smote:
                smote_resampler = SMOTE(random_state=42)
                X_train_resampled, y_train_resampled = smote_resampler.fit_resample(X_train, y_train)
                best_model.fit(X_train_resampled, y_train_resampled)
                X_train_for_predict = X_train_resampled # Use resampled data for train prediction
                y_train_for_predict = y_train_resampled
            else:
                best_model.fit(X_train, y_train)
                X_train_for_predict = X_train
                y_train_for_predict = y_train

            # Test predictions
            y_pred_test = best_model.predict(X_test)
            y_pred_test_total = np.append(y_pred_test_total, y_pred_test)
            y_true_test_total = np.append(y_true_test_total, y_test)

            # Train predictions
            y_pred_train = best_model.predict(X_train_for_predict)
            y_pred_train_total = np.append(y_pred_train_total, y_pred_train)
            y_true_train_total = np.append(y_true_train_total, y_train_for_predict)

        # Calculate test metrics
        test_accuracy = accuracy_score(y_true_test_total, y_pred_test_total)
        test_precision = precision_score(y_true_test_total, y_pred_test_total, average='binary', zero_division=0)
        test_recall = recall_score(y_true_test_total, y_pred_test_total, average='binary', zero_division=0)
        test_f1 = f1_score(y_true_test_total, y_pred_test_total, average='binary', zero_division=0)

        # Calculate train metrics
        train_accuracy = accuracy_score(y_true_train_total, y_pred_train_total)
        train_precision = precision_score(y_true_train_total, y_pred_train_total, average='binary', zero_division=0)
        train_recall = recall_score(y_true_train_total, y_pred_train_total, average='binary', zero_division=0)
        train_f1 = f1_score(y_true_train_total, y_pred_train_total, average='binary', zero_division=0)

        print(f"Train Accuracy: {train_accuracy:.4f}")
        print(f"Test Accuracy: {test_accuracy:.4f}")
        print(f"Train Precision: {train_precision:.4f}")
        print(f"Test Precision: {test_precision:.4f}")
        print(f"Train Recall: {train_recall:.4f}")
        print(f"Test Recall: {test_recall:.4f}")
        print(f"Train F1 Score: {train_f1:.4f}")
        print(f"Test F1 Score: {test_f1:.4f}")
        print("Test Classification Report:\n", classification_report(y_true_test_total, y_pred_test_total, zero_division=0))

        results[model_name] = {
            'best_model': best_model,
            'best_params': best_params,
            'train_accuracy': train_accuracy,
            'train_precision': train_precision,
            'train_recall': train_recall,
            'train_f1_score': train_f1,
            'accuracy': test_accuracy, # Renamed to 'test_accuracy' in summary
            'precision': test_precision, # Renamed to 'test_precision' in summary
            'recall': test_recall,       # Renamed to 'test_recall' in summary
            'f1_score': test_f1,         # Renamed to 'test_f1_score' in summary
            'test_classification_report': classification_report(y_true_test_total, y_pred_test_total, output_dict=True, zero_division=0)
        }
    return results

# Train and evaluate models for Cluster 0 (no SMOTE)
cluster0_results = train_evaluate_models(X0, y0, 'Cluster 0', use_smote=False)

# Train and evaluate models for Cluster 1 (with SMOTE)
cluster1_results = train_evaluate_models(X1, y1, 'Cluster 1', use_smote=True)
```

## Model Performance Summary

Let's review the performance of each model for both Cluster 0 and Cluster 1 based on the evaluation metrics (Accuracy, Precision, Recall, F1 Score) obtained from K-fold cross-validation.

```python
import pandas as pd

def summarize_results(results_dict, cluster_name):
    """Summarizes model performance results for a given cluster.

    Args:
        results_dict (dict): Dictionary containing results for each model.
        cluster_name (str): Name of the cluster.

    Returns:
        pd.DataFrame: A DataFrame summarizing model performance.
    """
    summary_data = []
    for model_name, metrics in results_dict.items():
        summary_data.append({
            'Model': model_name,
            'Train Accuracy': metrics['train_accuracy'],
            'Test Accuracy': metrics['accuracy'],
            'Train Precision': metrics['train_precision'],
            'Test Precision': metrics['precision'],
            'Train Recall': metrics['train_recall'],
            'Test Recall': metrics['recall'],
            'Train F1 Score': metrics['train_f1_score'],
            'Test F1 Score': metrics['f1_score'],
            'Best Params': metrics['best_params']
        })
    summary_df = pd.DataFrame(summary_data)
    summary_df = summary_df.sort_values(by='Test F1 Score', ascending=False).reset_index(drop=True)
    print(f"\n--- {cluster_name} Model Performance Summary ---")
    display(summary_df.round(4))
    return summary_df

# Summarize results for Cluster 0
summary_cluster0 = summarize_results(cluster0_results, 'Cluster 0')

# Summarize results for Cluster 1
summary_cluster1 = summarize_results(cluster1_results, 'Cluster 1')

print("\n--- Overall Best Model per Cluster ---")
print(f"Best Model for Cluster 0 (based on Test F1 Score): {summary_cluster0.iloc[0]['Model']} (Test F1 Score: {summary_cluster0.iloc[0]['Test F1 Score']:.4f})")
print(f"Best Model for Cluster 1 (based on Test F1 Score): {summary_cluster1.iloc[0]['Model']} (Test F1 Score: {summary_cluster1.iloc[0]['Test F1 Score']:.4f})")
```
