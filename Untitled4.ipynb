```python
# =========================================================
# COMPLETE FIXED CODE - HANDLING EXTREME CLASS IMBALANCE
# WITH PROPER ERROR HANDLING
# =========================================================

!pip install gower imbalanced-learn xgboost lightgbm

import pandas as pd
import numpy as np
import gower
import matplotlib.pyplot as plt
import seaborn as sns

from scipy.cluster.hierarchy import linkage, dendrogram, fcluster
from scipy.spatial.distance import squareform
from sklearn.metrics import silhouette_score, accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, classification_report, confusion_matrix
from sklearn.model_selection import StratifiedKFold, train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder, OrdinalEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.base import clone

# Advanced imbalanced learning techniques
from imblearn.over_sampling import SMOTE, ADASYN, BorderlineSMOTE
from imblearn.combine import SMOTEENN, SMOTETomek
from imblearn.under_sampling import RandomUnderSampler
from imblearn.pipeline import Pipeline as ImbPipeline

# Advanced classifiers for imbalanced data
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC

import warnings
warnings.filterwarnings('ignore')

# =========================================================
# 1. LOAD AND PREPARE DATA
# =========================================================

df = pd.read_csv('insurance2.csv')
df['sex'] = df['sex'].astype(bool)
df['smoker'] = df['smoker'].astype(bool)
df = df.drop_duplicates()

# Keep target as numeric for modeling
df['insuranceclaim_num'] = df['insuranceclaim'].copy()
df['insuranceclaim'] = df['insuranceclaim'].replace({0: 'No Claim', 1: 'Claim'})

print("Dataset shape:", df.shape)
print(f"Claim distribution:\n{df['insuranceclaim'].value_counts()}")
print(f"Claim rate: {df['insuranceclaim_num'].mean():.2%}")

# =========================================================
# 2. HIERARCHICAL CLUSTERING (YOUR ORIGINAL CODE - FIXED)
# =========================================================

numerical_cols = ['age', 'bmi', 'children', 'charges']
categorical_cols = ['sex', 'region', 'smoker']

# Prepare feature matrix
X_cluster = df[numerical_cols + categorical_cols].copy()

# Convert categorical to string for Gower
for col in categorical_cols:
    X_cluster[col] = X_cluster[col].astype(str)

# Categorical features indicator (True = categorical, False = numerical)
cat_features = [False] * len(numerical_cols) + [True] * len(categorical_cols)

print("\nComputing Gower distance matrix...")
gower_dist = gower.gower_matrix(X_cluster, cat_features=cat_features)
condensed_dist = squareform(gower_dist, checks=False)

# Hierarchical clustering
linked = linkage(condensed_dist, method='average')

# Silhouette analysis
silhouette_scores = []
K_range = range(2, 11)

for k in K_range:
    labels = fcluster(linked, k, criterion='maxclust')
    if len(np.unique(labels)) > 1:
        score = silhouette_score(gower_dist, labels, metric='precomputed')
        silhouette_scores.append(score)
    else:
        silhouette_scores.append(-1)

best_k = K_range[np.argmax(silhouette_scores)]
print(f"\nBest number of clusters: {best_k}")

# Assign clusters
df['Cluster'] = fcluster(linked, best_k, criterion='maxclust') - 1

# Verify cluster separation by smoker status
print("\nCluster vs Smoker cross-tabulation:")
print(pd.crosstab(df['Cluster'], df['smoker']))

# =========================================================
# 3. HANDLE IMBALANCED CLUSTERS WITH SPECIALIZED TECHNIQUES
# =========================================================

# Define multiple resampling strategies with proper error handling
def get_resampler_safe(resample_name, y_train):
    """Safely create resampler that won't fail with too few samples"""
    n_minority = min(np.bincount(y_train)) if len(np.unique(y_train)) > 1 else 0
    n_majority = max(np.bincount(y_train)) if len(np.unique(y_train)) > 1 else 0
    
    # If very few minority samples, use simpler methods
    if n_minority < 3:
        return None  # Will use class_weight instead
    
    resamplers = {
        'SMOTE': SMOTE(random_state=42, k_neighbors=min(n_minority-1, 3)),
        'SMOTE_ENN': SMOTEENN(random_state=42, smote=SMOTE(k_neighbors=min(n_minority-1, 3))),
        'SMOTE_Tomek': SMOTETomek(random_state=42, smote=SMOTE(k_neighbors=min(n_minority-1, 3))),
        'Random_Under': RandomUnderSampler(random_state=42)
    }
    
    # Only add ADASYN and Borderline if enough samples
    if n_minority >= 6:
        resamplers['ADASYN'] = ADASYN(random_state=42, n_neighbors=min(n_minority-1, 5))
        resamplers['Borderline_SMOTE'] = BorderlineSMOTE(random_state=42, k_neighbors=min(n_minority-1, 3))
    
    return resamplers.get(resample_name)

# Define classifiers optimized for imbalanced data
classifiers = {
    'XGBoost': XGBClassifier(random_state=42, scale_pos_weight=10, eval_metric='logloss'),
    'LightGBM': LGBMClassifier(random_state=42, class_weight='balanced', verbose=-1),
    'RandomForest': RandomForestClassifier(random_state=42, class_weight='balanced'),
    'GradientBoosting': GradientBoostingClassifier(random_state=42)
}

# Preprocessor for modeling
def create_preprocessor():
    numerical_features = ['age', 'bmi', 'children', 'charges']
    ordinal_features = ['sex', 'smoker']
    onehot_features = ['region']
    
    return ColumnTransformer(
        transformers=[
            ('num', StandardScaler(), numerical_features),
            ('ord', OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1), ordinal_features),
            ('ohe', OneHotEncoder(handle_unknown='ignore', drop='first'), onehot_features)
        ]
    )

# =========================================================
# 4. RARE EVENT MODELING FOR EACH CLUSTER
# =========================================================

cluster_results = {}

for cluster_id in sorted(df['Cluster'].unique()):
    print(f"\n{'='*60}")
    print(f"PROCESSING CLUSTER {cluster_id}")
    print(f"{'='*60}")
    
    # Filter data for this cluster
    cluster_data = df[df['Cluster'] == cluster_id].copy()
    X = cluster_data[numerical_cols + categorical_cols]
    y = cluster_data['insuranceclaim_num']
    
    # Check class distribution
    n_claims = y.sum()
    n_no_claims = len(y) - n_claims
    claim_rate = n_claims / len(y)
    
    print(f"Cluster size: {len(y)}")
    print(f"Claims: {n_claims} ({claim_rate:.2%})")
    print(f"No claims: {n_no_claims} ({1-claim_rate:.2%})")
    
    # Skip if only one class present
    if len(np.unique(y)) < 2:
        print(f"\n⚠️ SKIPPING: Only one class present in Cluster {cluster_id}")
        cluster_results[f'Cluster_{cluster_id}'] = {
            'cluster_size': len(y),
            'claim_rate': claim_rate,
            'technique': 'Only one class',
            'accuracy': 1.0 if n_claims == 0 else 0,
            'precision': 0,
            'recall': 0,
            'f1_score': 0,
            'roc_auc': 0
        }
        continue
    
    # Determine which technique to use based on imbalance severity
    if claim_rate < 0.05 or n_claims < 10:
        print(f"\n⚠️ SEVERE IMBALANCE DETECTED in Cluster {cluster_id}")
        print(f"Using Weighted Class Approach instead of SMOTE")
        
        # =========================================================
        # WEIGHTED CLASS APPROACH FOR EXTREME IMBALANCE
        # =========================================================
        
        try:
            # Calculate class weight
            class_weight = {0: 1.0, 1: n_no_claims / n_claims}
            
            # Train model with class weights
            preprocessor = create_preprocessor()
            model = XGBClassifier(
                random_state=42, 
                scale_pos_weight=n_no_claims / n_claims,
                eval_metric='logloss'
            )
            pipeline = Pipeline([('preprocessor', preprocessor), ('classifier', model)])
            
            # Use stratified split
            X_train, X_test, y_train, y_test = train_test_split(
                X, y, test_size=0.3, stratify=y, random_state=42
            )
            
            pipeline.fit(X_train, y_train)
            
            # Evaluate
            y_pred = pipeline.predict(X_test)
            y_prob = pipeline.predict_proba(X_test)[:, 1]
            
            results = {
                'cluster_size': len(y),
                'claim_rate': claim_rate,
                'technique': 'Weighted_XGBoost',
                'accuracy': accuracy_score(y_test, y_pred),
                'precision': precision_score(y_test, y_pred, zero_division=0),
                'recall': recall_score(y_test, y_pred, zero_division=0),
                'f1_score': f1_score(y_test, y_pred, zero_division=0),
                'roc_auc': roc_auc_score(y_test, y_prob)
            }
            
            cluster_results[f'Cluster_{cluster_id}'] = results
            
            print(f"\nResults for Cluster {cluster_id} (Weighted XGBoost):")
            for metric, value in results.items():
                if metric not in ['cluster_size', 'claim_rate', 'technique']:
                    print(f"  {metric}: {value:.4f}")
                    
        except Exception as e:
            print(f"Error with weighted approach: {e}")
            # Fallback to simple model
            preprocessor = create_preprocessor()
            model = RandomForestClassifier(random_state=42, class_weight='balanced')
            pipeline = Pipeline([('preprocessor', preprocessor), ('classifier', model)])
            
            X_train, X_test, y_train, y_test = train_test_split(
                X, y, test_size=0.3, stratify=y, random_state=42
            )
            pipeline.fit(X_train, y_train)
            y_pred = pipeline.predict(X_test)
            y_prob = pipeline.predict_proba(X_test)[:, 1]
            
            results = {
                'cluster_size': len(y),
                'claim_rate': claim_rate,
                'technique': 'Balanced_RF_Fallback',
                'accuracy': accuracy_score(y_test, y_pred),
                'precision': precision_score(y_test, y_pred, zero_division=0),
                'recall': recall_score(y_test, y_pred, zero_division=0),
                'f1_score': f1_score(y_test, y_pred, zero_division=0),
                'roc_auc': roc_auc_score(y_test, y_prob)
            }
            cluster_results[f'Cluster_{cluster_id}'] = results
        
    else:
        # For clusters with less severe imbalance, try resampling strategies
        best_results = {'best_f1': 0, 'best_strategy': None}
        
        # List of strategies to try (safe ones only)
        safe_strategies = ['SMOTE', 'Random_Under']
        if n_claims >= 6:
            safe_strategies.extend(['SMOTE_ENN', 'SMOTE_Tomek'])
        
        for resample_name in safe_strategies:
            print(f"\n  Trying {resample_name}...")
            
            try:
                # Create resampler safely
                resampler = get_resampler_safe(resample_name, y)
                if resampler is None:
                    print(f"    Skipping {resample_name} - insufficient samples")
                    continue
                
                # Create pipeline with resampling
                preprocessor = create_preprocessor()
                model = XGBClassifier(random_state=42, scale_pos_weight=5, eval_metric='logloss')
                
                pipeline = ImbPipeline([
                    ('preprocessor', preprocessor),
                    ('resampler', resampler),
                    ('classifier', model)
                ])
                
                # Cross-validation for imbalanced data
                skf = StratifiedKFold(n_splits=min(3, len(np.unique(y))), shuffle=True, random_state=42)
                cv_results = []
                
                for train_idx, test_idx in skf.split(X, y):
                    X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
                    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]
                    
                    # Ensure at least 2 classes in training
                    if len(np.unique(y_train)) < 2:
                        continue
                    
                    pipeline_clone = clone(pipeline)
                    pipeline_clone.fit(X_train, y_train)
                    
                    y_pred = pipeline_clone.predict(X_test)
                    y_prob = pipeline_clone.predict_proba(X_test)[:, 1]
                    
                    cv_results.append({
                        'accuracy': accuracy_score(y_test, y_pred),
                        'precision': precision_score(y_test, y_pred, zero_division=0),
                        'recall': recall_score(y_test, y_pred, zero_division=0),
                        'f1': f1_score(y_test, y_pred, zero_division=0),
                        'roc_auc': roc_auc_score(y_test, y_prob)
                    })
                
                if cv_results:
                    avg_f1 = np.mean([r['f1'] for r in cv_results])
                    
                    if avg_f1 > best_results['best_f1']:
                        best_results['best_f1'] = avg_f1
                        best_results['best_strategy'] = resample_name
                        best_results['results'] = {
                            'accuracy': np.mean([r['accuracy'] for r in cv_results]),
                            'precision': np.mean([r['precision'] for r in cv_results]),
                            'recall': np.mean([r['recall'] for r in cv_results]),
                            'f1_score': avg_f1,
                            'roc_auc': np.mean([r['roc_auc'] for r in cv_results])
                        }
                
            except Exception as e:
                print(f"    Error with {resample_name}: {str(e)[:100]}")
                continue
        
        if best_results.get('results'):
            cluster_results[f'Cluster_{cluster_id}'] = {
                'cluster_size': len(y),
                'claim_rate': claim_rate,
                'technique': best_results['best_strategy'],
                **best_results['results']
            }
            
            print(f"\n✅ Best for Cluster {cluster_id}: {best_results['best_strategy']}")
            print(f"   F1-Score: {best_results['best_f1']:.4f}")
        else:
            # Fallback to simple weighted model
            print(f"\n⚠️ No resampling worked for Cluster {cluster_id}, using weighted approach")
            preprocessor = create_preprocessor()
            model = XGBClassifier(random_state=42, scale_pos_weight=n_no_claims/n_claims, eval_metric='logloss')
            pipeline = Pipeline([('preprocessor', preprocessor), ('classifier', model)])
            
            skf = StratifiedKFold(n_splits=min(3, len(np.unique(y))), shuffle=True, random_state=42)
            cv_results = []
            
            for train_idx, test_idx in skf.split(X, y):
                X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
                y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]
                
                if len(np.unique(y_train)) < 2:
                    continue
                
                pipeline_clone = clone(pipeline)
                pipeline_clone.fit(X_train, y_train)
                
                y_pred = pipeline_clone.predict(X_test)
                y_prob = pipeline_clone.predict_proba(X_test)[:, 1]
                
                cv_results.append({
                    'accuracy': accuracy_score(y_test, y_pred),
                    'precision': precision_score(y_test, y_pred, zero_division=0),
                    'recall': recall_score(y_test, y_pred, zero_division=0),
                    'f1': f1_score(y_test, y_pred, zero_division=0),
                    'roc_auc': roc_auc_score(y_test, y_prob)
                })
            
            if cv_results:
                cluster_results[f'Cluster_{cluster_id}'] = {
                    'cluster_size': len(y),
                    'claim_rate': claim_rate,
                    'technique': 'Weighted_XGBoost_Fallback',
                    'accuracy': np.mean([r['accuracy'] for r in cv_results]),
                    'precision': np.mean([r['precision'] for r in cv_results]),
                    'recall': np.mean([r['recall'] for r in cv_results]),
                    'f1_score': np.mean([r['f1'] for r in cv_results]),
                    'roc_auc': np.mean([r['roc_auc'] for r in cv_results])
                }

# =========================================================
# 5. DISPLAY RESULTS
# =========================================================

if cluster_results:
    cluster_df = pd.DataFrame(cluster_results).T
    cluster_df = cluster_df.round(4)
    
    print("\n" + "="*80)
    print("FINAL RESULTS")
    print("="*80)
    
    # Display results in a clean format
    display_cols = ['cluster_size', 'claim_rate', 'technique', 'f1_score', 'recall', 'precision', 'roc_auc']
    available_cols = [col for col in display_cols if col in cluster_df.columns]
    
    print(cluster_df[available_cols])
    
    # Create summary visualization
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    
    # Plot 1: F1-Score by cluster
    ax1 = axes[0]
    if 'f1_score' in cluster_df.columns:
        clusters = cluster_df.index
        f1_scores = cluster_df['f1_score']
        colors = ['red' if 'Weighted' in str(cluster_df.loc[c, 'technique']) else 'green' for c in clusters]
        ax1.bar(range(len(clusters)), f1_scores, color=colors, alpha=0.7)
        ax1.set_xticks(range(len(clusters)))
        ax1.set_xticklabels(clusters, rotation=45)
        ax1.set_ylabel('F1-Score')
        ax1.set_title('F1-Score by Cluster')
        ax1.set_ylim([0, 1])
        ax1.grid(alpha=0.3, axis='y')
        
    # Plot 2: Technique used
    ax2 = axes[1]
    if 'technique' in cluster_df.columns:
        technique_counts = cluster_df['technique'].value_counts()
        ax2.pie(technique_counts.values, labels=technique_counts.index, autopct='%1.1f%%')
        ax2.set_title('Techniques Used per Cluster')
    
    plt.tight_layout()
    plt.show()
    
    # Print insights
    print("\n" + "="*80)
    print("KEY INSIGHTS")
    print("="*80)
    
    for cluster_id in sorted(df['Cluster'].unique()):
        cluster_data = df[df['Cluster'] == cluster_id]
        if 'insuranceclaim_num' in cluster_data.columns:
            claim_rate = cluster_data['insuranceclaim_num'].mean() * 100
            smokers = cluster_data['smoker'].sum() if 'smoker' in cluster_data.columns else 0
            smoker_pct = smokers / len(cluster_data) * 100
            print(f"\nCluster {cluster_id}:")
            print(f"  - Size: {len(cluster_data)} samples")
            print(f"  - Claim rate: {claim_rate:.1f}%")
            print(f"  - Smoker rate: {smoker_pct:.1f}%")
            if f'Cluster_{cluster_id}' in cluster_results:
                print(f"  - Best technique: {cluster_results[f'Cluster_{cluster_id}']['technique']}")
                print(f"  - F1-Score: {cluster_results[f'Cluster_{cluster_id}']['f1_score']:.4f}")
    
    # Save results
    cluster_df.to_csv('cluster_analysis_results.csv')
    print("\n✓ Results saved to 'cluster_analysis_results.csv'")
    
else:
    print("\n⚠️ No results generated. Please check your data.")
```

```python
# =========================================================
# CLUSTER 0 MODEL BUILDING AND COMPARISON
# BALANCED DATASET (SMOKERS) - NO BALANCING TECHNIQUES
# =========================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split, cross_val_score, StratifiedKFold
from sklearn.preprocessing import StandardScaler, OneHotEncoder, OrdinalEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, confusion_matrix, classification_report

# Classifiers - standard models without any balancing
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier

import warnings
warnings.filterwarnings('ignore')

# =========================================================
# 1. LOAD DATA AND FILTER CLUSTER 0
# =========================================================

# Load data
df = pd.read_csv('insurance2.csv')

# Convert boolean columns
df['sex'] = df['sex'].astype(bool)
df['smoker'] = df['smoker'].astype(bool)

# Remove duplicates
df = df.drop_duplicates()

# Create numeric target
df['insuranceclaim_num'] = df['insuranceclaim'].copy()

# =========================================================
# 2. IDENTIFY CLUSTER 0 (Based on your Gower clustering)
# =========================================================

if 'Cluster' in df.columns:
    cluster1_data = df[df['Cluster'] == 1].copy()
else:
    # Create based on the pattern (Cluster 1 = non-smokers as per your discovery)
    df['Cluster'] = 0
    df.loc[df['smoker'] == False, 'Cluster'] = 1
    cluster1_data = df[df['Cluster'] == 1].copy()

# For this code, we're assuming Cluster 0 has already been identified
# Based on your analysis, Cluster 0 = all smokers


# Define features
numerical_cols = ['age', 'bmi', 'children', 'charges']
categorical_cols = ['sex', 'region', 'smoker']

X = cluster0_data[numerical_cols + categorical_cols]
y = cluster0_data['insuranceclaim_num']

# =========================================================
# 3. VERIFY CLUSTER 0 IS BALANCED
# =========================================================

print("="*80)
print("CLUSTER 0 VERIFICATION")
print("="*80)
print(f"Cluster 0 Size: {len(X)} samples")
print(f"Claims (1): {y.sum()} ({y.mean():.2%})")
print(f"No Claims (0): {len(y) - y.sum()} ({1-y.mean():.2%})")
print(f"Status: {'✓ BALANCED' if 0.4 <= y.mean() <= 0.6 else '✗ IMBALANCED'}")
print(f"Smokers in Cluster 0: {cluster0_data['smoker'].sum()} ({cluster0_data['smoker'].mean()*100:.0f}%)")
print("\n⚠️ CONFIRMED: NO BALANCING TECHNIQUES WILL BE USED")

# =========================================================
# 4. CREATE PREPROCESSOR (Standard scaling for numerical, encoding for categorical)
# =========================================================

def create_preprocessor():
    numerical_features = ['age', 'bmi', 'children', 'charges']
    ordinal_features = ['sex', 'smoker']
    onehot_features = ['region']
    
    return ColumnTransformer(
        transformers=[
            ('num', StandardScaler(), numerical_features),
            ('ord', OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1), ordinal_features),
            ('ohe', OneHotEncoder(handle_unknown='ignore', drop='first'), onehot_features)
        ]
    )

# =========================================================
# 5. DEFINE MODELS (NO BALANCING PARAMETERS)
# =========================================================

models = {
    'Logistic Regression': LogisticRegression(random_state=42, max_iter=1000),
    'Decision Tree': DecisionTreeClassifier(random_state=42),
    'Random Forest': RandomForestClassifier(random_state=42, n_estimators=100),
    'Gradient Boosting': GradientBoostingClassifier(random_state=42, n_estimators=100),
    'XGBoost': XGBClassifier(random_state=42, eval_metric='logloss', use_label_encoder=False),
    'LightGBM': LGBMClassifier(random_state=42, verbose=-1),
    'SVM': SVC(random_state=42, probability=True),
    'KNN': KNeighborsClassifier(n_neighbors=5)
}

# =========================================================
# 6. TRAIN AND EVALUATE MODELS (PURE NORMAL TRAINING)
# =========================================================

results = []

print("\n" + "="*80)
print("MODEL TRAINING ON CLUSTER 0 (NO BALANCING TECHNIQUES)")
print("="*80)

for model_name, model in models.items():
    print(f"\n▶ Training {model_name}...")
    
    # Create pipeline - NO balancing whatsoever
    preprocessor = create_preprocessor()
    pipeline = Pipeline([
        ('preprocessor', preprocessor),
        ('classifier', model)  # Direct model - no class weights, no SMOTE
    ])
    
    # Split data (stratified to maintain class balance)
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=0.3, random_state=42, stratify=y
    )
    
    # Train normally
    pipeline.fit(X_train, y_train)
    
    # Predict
    y_train_pred = pipeline.predict(X_train)
    y_test_pred = pipeline.predict(X_test)
    
    # Get probabilities for ROC-AUC
    try:
        y_test_prob = pipeline.predict_proba(X_test)[:, 1]
        y_train_prob = pipeline.predict_proba(X_train)[:, 1]
    except:
        y_test_prob = np.zeros_like(y_test)
        y_train_prob = np.zeros_like(y_train)
    
    # Calculate metrics
    train_acc = accuracy_score(y_train, y_train_pred)
    test_acc = accuracy_score(y_test, y_test_pred)
    precision = precision_score(y_test, y_test_pred, zero_division=0)
    recall = recall_score(y_test, y_test_pred, zero_division=0)
    f1 = f1_score(y_test, y_test_pred, zero_division=0)
    train_roc = roc_auc_score(y_train, y_train_prob) if len(np.unique(y_train)) > 1 else 0
    test_roc = roc_auc_score(y_test, y_test_prob) if len(np.unique(y_test)) > 1 else 0
    
    # Store results
    results.append({
        'Model': model_name,
        'Train_Accuracy': train_acc,
        'Test_Accuracy': test_acc,
        'Train_ROC_AUC': train_roc,
        'Test_ROC_AUC': test_roc,
        'Precision': precision,
        'Recall': recall,
        'F1_Score': f1,
        'Overfitting_Gap': train_acc - test_acc
    })
    
    print(f"  Train Acc: {train_acc:.4f} | Test Acc: {test_acc:.4f} | F1: {f1:.4f} | Gap: {train_acc - test_acc:.4f}")

# =========================================================
# 7. CREATE RESULTS DATAFRAME
# =========================================================

results_df = pd.DataFrame(results)
results_df = results_df.sort_values('Test_Accuracy', ascending=False)

print("\n" + "="*80)
print("CLUSTER 0 - MODEL COMPARISON TABLE")
print("(NO BALANCING TECHNIQUES USED)")
print("="*80)
print(results_df[['Model', 'Train_Accuracy', 'Test_Accuracy', 'Precision', 'Recall', 'F1_Score', 'Test_ROC_AUC', 'Overfitting_Gap']].to_string(index=False))

# =========================================================
# 8. SUMMARY STATISTICS
# =========================================================

print("\n" + "="*80)
print("SUMMARY STATISTICS")
print("="*80)
print(f"Best Model by Test Accuracy: {results_df.iloc[0]['Model']} ({results_df.iloc[0]['Test_Accuracy']:.4f})")
print(f"Best Model by F1-Score: {results_df.loc[results_df['F1_Score'].idxmax(), 'Model']} ({results_df['F1_Score'].max():.4f})")
print(f"Average Test Accuracy: {results_df['Test_Accuracy'].mean():.4f}")
print(f"Average F1-Score: {results_df['F1_Score'].mean():.4f}")

# Overfitting analysis
high_gap = results_df[results_df['Overfitting_Gap'] > 0.05]
if len(high_gap) > 0:
    print(f"\n⚠️ Models showing overfitting (gap > 0.05):")
    for _, row in high_gap.iterrows():
        print(f"  - {row['Model']}: Gap = {row['Overfitting_Gap']:.4f}")
else:
    print("\n✓ No significant overfitting detected in any model")

# =========================================================
# 9. CONFUSION MATRIX FOR BEST MODEL
# =========================================================

best_model_name = results_df.iloc[0]['Model']
best_model = models[best_model_name]

# Retrain best model on full data to get confusion matrix
preprocessor = create_preprocessor()
pipeline = Pipeline([
    ('preprocessor', preprocessor),
    ('classifier', best_model)
])

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)
pipeline.fit(X_train, y_train)
y_pred = pipeline.predict(X_test)

cm = confusion_matrix(y_test, y_pred)

print("\n" + "="*80)
print(f"CONFUSION MATRIX - BEST MODEL: {best_model_name}")
print("="*80)
print(f"                 Predicted")
print(f"                 No Claim  Claim")
print(f"Actual No Claim  {cm[0,0]:4d}    {cm[0,1]:4d}")
print(f"       Claim      {cm[1,0]:4d}    {cm[1,1]:4d}")
print(f"\nClassification Report:")
print(classification_report(y_test, y_pred, target_names=['No Claim', 'Claim']))

# =========================================================
# 10. VISUALIZATIONS
# =========================================================

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Plot 1: Model Performance Comparison
ax1 = axes[0, 0]
models_plot = results_df['Model'].tolist()
x = np.arange(len(models_plot))
width = 0.35

ax1.bar(x - width/2, results_df['Train_Accuracy'], width, label='Train Accuracy', alpha=0.8, color='blue')
ax1.bar(x + width/2, results_df['Test_Accuracy'], width, label='Test Accuracy', alpha=0.8, color='green')
ax1.set_xlabel('Models')
ax1.set_ylabel('Accuracy')
ax1.set_title('Cluster 0: Train vs Test Accuracy (NO BALANCING)')
ax1.set_xticks(x)
ax1.set_xticklabels(models_plot, rotation=45, ha='right')
ax1.legend()
ax1.grid(alpha=0.3, axis='y')
ax1.set_ylim([0, 1])

# Plot 2: F1-Score and ROC-AUC
ax2 = axes[0, 1]
ax2.bar(x - width/2, results_df['F1_Score'], width, label='F1-Score', alpha=0.8, color='orange')
ax2.bar(x + width/2, results_df['Test_ROC_AUC'], width, label='ROC-AUC', alpha=0.8, color='red')
ax2.set_xlabel('Models')
ax2.set_ylabel('Score')
ax2.set_title('Cluster 0: F1-Score and ROC-AUC')
ax2.set_xticks(x)
ax2.set_xticklabels(models_plot, rotation=45, ha='right')
ax2.legend()
ax2.grid(alpha=0.3, axis='y')
ax2.set_ylim([0, 1])

# Plot 3: Overfitting Gap
ax3 = axes[1, 0]
colors = ['red' if gap > 0.05 else 'green' for gap in results_df['Overfitting_Gap']]
ax3.barh(models_plot, results_df['Overfitting_Gap'], color=colors, alpha=0.7)
ax3.set_xlabel('Overfitting Gap (Train - Test Accuracy)')
ax3.set_title('Cluster 0: Overfitting Analysis')
ax3.axvline(x=0.05, color='red', linestyle='--', label='Overfitting threshold (0.05)')
ax3.legend()
ax3.grid(alpha=0.3, axis='x')

# Plot 4: Confusion Matrix Heatmap for Best Model
ax4 = axes[1, 1]
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=ax4,
            xticklabels=['No Claim', 'Claim'],
            yticklabels=['No Claim', 'Claim'])
ax4.set_title(f'Confusion Matrix - Best Model: {best_model_name}')
ax4.set_ylabel('Actual')
ax4.set_xlabel('Predicted')

plt.tight_layout()
plt.show()

# =========================================================
# 11. CROSS-VALIDATION FOR BEST MODEL
# =========================================================

print("\n" + "="*80)
print(f"CROSS-VALIDATION FOR BEST MODEL: {best_model_name}")
print("="*80)

# Perform 5-fold cross-validation on best model
preprocessor = create_preprocessor()
best_pipeline = Pipeline([
    ('preprocessor', preprocessor),
    ('classifier', best_model)
])

cv_scores = cross_val_score(best_pipeline, X, y, cv=5, scoring='accuracy')
cv_f1 = cross_val_score(best_pipeline, X, y, cv=5, scoring='f1_macro')

print(f"5-Fold CV Accuracy: {cv_scores}")
print(f"Mean CV Accuracy: {cv_scores.mean():.4f} (+/- {cv_scores.std() * 2:.4f})")
print(f"5-Fold CV F1-Score: {cv_f1.mean():.4f} (+/- {cv_f1.std() * 2:.4f})")

# =========================================================
# 12. SAVE RESULTS
# =========================================================

results_df.to_csv('cluster0_results_NO_balancing.csv', index=False)
print("\n" + "="*80)
print("RESULTS SAVED: cluster0_results_NO_balancing.csv")
print("="*80)
print("\n✅ COMPLETE: No balancing techniques were used for Cluster 0")
print("✅ All models trained on original, balanced data")
print("✅ Comparison table shows train and test accuracies for all models")
```

```python
# =========================================================
# CODE 2: CLUSTER 1 MODEL BUILDING
# IMBALANCED DATASET (25 No Claim, 249 Claim)
# WITH SMOTE BALANCING TECHNIQUES
# =========================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split, cross_val_score, StratifiedKFold
from sklearn.preprocessing import StandardScaler, OneHotEncoder, OrdinalEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, confusion_matrix, classification_report

# Classifiers
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier

# Imbalanced learning techniques
from imblearn.over_sampling import SMOTE, ADASYN, BorderlineSMOTE
from imblearn.combine import SMOTEENN, SMOTETomek
from imblearn.pipeline import Pipeline as ImbPipeline

import warnings
warnings.filterwarnings('ignore')

# =========================================================
# 1. LOAD DATA AND FILTER CLUSTER 1
# =========================================================

# Load data
df = pd.read_csv('insurance2.csv')

# Convert boolean columns
df['sex'] = df['sex'].astype(bool)
df['smoker'] = df['smoker'].astype(bool)

# Remove duplicates
df = df.drop_duplicates()

# Create numeric target
df['insuranceclaim_num'] = df['insuranceclaim'].copy()

# Based on your cluster profile, assign clusters
# Cluster 1: 25 No Claim, 249 Claim (Highly Imbalanced)
df['Cluster'] = 0  # Temporary - replace with actual cluster assignment
df.loc[df['smoker'] == False, 'Cluster'] = 1  # Non-smokers go to Cluster 1

# Filter ONLY Cluster 0
cluster0_data = df[df['Cluster'] == 0].copy()

# Define features
numerical_cols = ['age', 'bmi', 'children', 'charges']
categorical_cols = ['sex', 'region', 'smoker']

X = cluster1_data[numerical_cols + categorical_cols]
y = cluster1_data['insuranceclaim_num']

# =========================================================
# 2. VERIFY CLUSTER 1 IMBALANCE
# =========================================================

print("="*80)
print("CLUSTER 1 VERIFICATION")
print("="*80)
print(f"Cluster 1 Size: {len(X)} samples")
print(f"No Claims (0): {(y == 0).sum()} ({(y == 0).mean():.2%})")
print(f"Claims (1): {y.sum()} ({y.mean():.2%})")
print(f"Imbalance Ratio: {(y == 0).sum() / y.sum():.2f}:1")
print(f"\n⚠️ CONFIRMATION: IMBALANCED DATASET")
print("   - WILL USE SMOTE AND OTHER BALANCING TECHNIQUES")
print("   - Minority class (No Claims) has only 25 samples")

# =========================================================
# 3. CREATE PREPROCESSOR
# =========================================================

def create_preprocessor():
    numerical_features = ['age', 'bmi', 'children', 'charges']
    ordinal_features = ['sex', 'smoker']
    onehot_features = ['region']
    
    return ColumnTransformer(
        transformers=[
            ('num', StandardScaler(), numerical_features),
            ('ord', OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1), ordinal_features),
            ('ohe', OneHotEncoder(handle_unknown='ignore', drop='first'), onehot_features)
        ]
    )

# =========================================================
# 4. DEFINE MODELS AND SMOTE STRATEGIES
# =========================================================

models = {
    'Logistic Regression': LogisticRegression(random_state=42, max_iter=1000),
    'Decision Tree': DecisionTreeClassifier(random_state=42),
    'Random Forest': RandomForestClassifier(random_state=42, n_estimators=100),
    'Gradient Boosting': GradientBoostingClassifier(random_state=42, n_estimators=100),
    'XGBoost': XGBClassifier(random_state=42, eval_metric='logloss', use_label_encoder=False),
    'LightGBM': LGBMClassifier(random_state=42, verbose=-1),
    'SVM': SVC(random_state=42, probability=True),
    'KNN': KNeighborsClassifier(n_neighbors=5)
}

# Define SMOTE and other balancing strategies
balancing_strategies = {
    'SMOTE': SMOTE(random_state=42),
    'ADASYN': ADASYN(random_state=42),
    'Borderline_SMOTE': BorderlineSMOTE(random_state=42),
    'SMOTE_ENN': SMOTEENN(random_state=42),
    'SMOTE_Tomek': SMOTETomek(random_state=42)
}

# =========================================================
# 5. TRAIN MODELS WITH DIFFERENT BALANCING STRATEGIES
# =========================================================

all_results = []

print("\n" + "="*80)
print("CLUSTER 1 - TRAINING WITH SMOTE BALANCING TECHNIQUES")
print("="*80)

for model_name, model in models.items():
    print(f"\n▶ Testing {model_name} with different SMOTE strategies...")
    
    best_f1 = 0
    best_strategy = None
    best_metrics = None
    
    # Try different balancing strategies
    for strategy_name, smote in balancing_strategies.items():
        try:
            # Create pipeline with SMOTE
            preprocessor = create_preprocessor()
            pipeline = ImbPipeline([
                ('preprocessor', preprocessor),
                ('smote', smote),
                ('classifier', model)
            ])
            
            # Split data with stratification
            X_train, X_test, y_train, y_test = train_test_split(
                X, y, test_size=0.3, random_state=42, stratify=y
            )
            
            # Train
            pipeline.fit(X_train, y_train)
            
            # Predict
            y_pred = pipeline.predict(X_test)
            y_prob = pipeline.predict_proba(X_test)[:, 1]
            
            # Calculate metrics
            test_acc = accuracy_score(y_test, y_pred)
            precision = precision_score(y_test, y_pred, zero_division=0)
            recall = recall_score(y_test, y_pred, zero_division=0)
            f1 = f1_score(y_test, y_pred, zero_division=0)
            roc_auc = roc_auc_score(y_test, y_prob)
            
            # Track best based on F1-score (better for imbalanced)
            if f1 > best_f1:
                best_f1 = f1
                best_strategy = strategy_name
                best_metrics = {
                    'Model': model_name,
                    'SMOTE_Strategy': strategy_name,
                    'Test_Accuracy': test_acc,
                    'Precision': precision,
                    'Recall': recall,
                    'F1_Score': f1,
                    'ROC_AUC': roc_auc
                }
                
        except Exception as e:
            print(f"    {strategy_name} failed: {str(e)[:50]}")
            continue
    
    if best_metrics:
        # Also get training accuracy for best strategy
        preprocessor = create_preprocessor()
        pipeline = ImbPipeline([
            ('preprocessor', preprocessor),
            ('smote', balancing_strategies[best_strategy]),
            ('classifier', model)
        ])
        
        X_train, X_test, y_train, y_test = train_test_split(
            X, y, test_size=0.3, random_state=42, stratify=y
        )
        pipeline.fit(X_train, y_train)
        y_train_pred = pipeline.predict(X_train)
        train_acc = accuracy_score(y_train, y_train_pred)
        
        best_metrics['Train_Accuracy'] = train_acc
        best_metrics['Overfitting_Gap'] = train_acc - best_metrics['Test_Accuracy']
        all_results.append(best_metrics)
        
        print(f"  ✓ BEST: {best_strategy} | Test Acc: {best_metrics['Test_Accuracy']:.4f} | F1: {best_metrics['F1_Score']:.4f} | Recall: {best_metrics['Recall']:.4f}")

# =========================================================
# 6. RESULTS TABLE
# =========================================================

results_df = pd.DataFrame(all_results)
results_df = results_df.sort_values('F1_Score', ascending=False)

print("\n" + "="*80)
print("CLUSTER 1 - MODEL COMPARISON TABLE")
print("(WITH SMOTE BALANCING TECHNIQUES)")
print("="*80)
print(results_df[['Model', 'SMOTE_Strategy', 'Train_Accuracy', 'Test_Accuracy', 'Precision', 'Recall', 'F1_Score', 'ROC_AUC', 'Overfitting_Gap']].to_string(index=False))

# =========================================================
# 7. BEST MODEL DETAILS
# =========================================================

best_model_row = results_df.iloc[0]
best_model_name = best_model_row['Model']
best_strategy = best_model_row['SMOTE_Strategy']
best_model = models[best_model_name]
best_smote = balancing_strategies[best_strategy]

print("\n" + "="*80)
print(f"BEST MODEL FOR CLUSTER 1: {best_model_name} with {best_strategy}")
print("="*80)
print(f"Test Accuracy: {best_model_row['Test_Accuracy']:.4f}")
print(f"F1-Score: {best_model_row['F1_Score']:.4f}")
print(f"Recall: {best_model_row['Recall']:.4f}")
print(f"ROC-AUC: {best_model_row['ROC_AUC']:.4f}")

# =========================================================
# 8. CONFUSION MATRIX FOR BEST MODEL
# =========================================================

preprocessor = create_preprocessor()
pipeline = ImbPipeline([
    ('preprocessor', preprocessor),
    ('smote', best_smote),
    ('classifier', best_model)
])

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)
pipeline.fit(X_train, y_train)
y_pred = pipeline.predict(X_test)
y_prob = pipeline.predict_proba(X_test)[:, 1]

cm = confusion_matrix(y_test, y_pred)

print("\n" + "="*80)
print(f"CONFUSION MATRIX - {best_model_name} with {best_strategy}")
print("="*80)
print(f"\n                 Predicted")
print(f"                 No Claim  Claim")
print(f"Actual No Claim  {cm[0,0]:4d}    {cm[0,1]:4d}")
print(f"       Claim      {cm[1,0]:4d}    {cm[1,1]:4d}")

print(f"\nClassification Report:")
print(classification_report(y_test, y_pred, target_names=['No Claim', 'Claim']))

# =========================================================
# 9. VISUALIZATIONS
# =========================================================

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Plot 1: F1-Score Comparison (most important for imbalanced)
ax1 = axes[0, 0]
models_list = results_df['Model'].tolist()
f1_scores = results_df['F1_Score'].tolist()
colors = ['green' if f1 > 0.7 else 'orange' if f1 > 0.5 else 'red' for f1 in f1_scores]
ax1.barh(models_list, f1_scores, color=colors, alpha=0.7)
ax1.set_xlabel('F1-Score')
ax1.set_title('Cluster 1: F1-Score by Model (WITH SMOTE)')
ax1.set_xlim([0, 1])
ax1.grid(alpha=0.3, axis='x')

# Plot 2: Recall vs Precision
ax2 = axes[0, 1]
x = np.arange(len(models_list))
width = 0.35
ax2.barh(x - width/2, results_df['Recall'], width, label='Recall', alpha=0.7, color='blue')
ax2.barh(x + width/2, results_df['Precision'], width, label='Precision', alpha=0.7, color='orange')
ax2.set_yticks(x)
ax2.set_yticklabels(models_list)
ax2.set_xlabel('Score')
ax2.set_title('Cluster 1: Recall vs Precision (WITH SMOTE)')
ax2.legend()
ax2.set_xlim([0, 1])
ax2.grid(alpha=0.3, axis='x')

# Plot 3: SMOTE Strategy Distribution
ax3 = axes[1, 0]
strategy_counts = results_df['SMOTE_Strategy'].value_counts()
ax3.pie(strategy_counts.values, labels=strategy_counts.index, autopct='%1.1f%%')
ax3.set_title('Best SMOTE Strategies for Cluster 1')

# Plot 4: Confusion Matrix Heatmap
ax4 = axes[1, 1]
sns.heatmap(cm, annot=True, fmt='d', cmap='Reds', ax=ax4,
            xticklabels=['No Claim', 'Claim'],
            yticklabels=['No Claim', 'Claim'])
ax4.set_title(f'Confusion Matrix - {best_model_name} with {best_strategy}')
ax4.set_ylabel('Actual')
ax4.set_xlabel('Predicted')

plt.tight_layout()
plt.show()

# =========================================================
# 10. SAVE RESULTS
# =========================================================

results_df.to_csv('cluster1_imbalanced_WITH_SMOTE.csv', index=False)
print("\n" + "="*80)
print("RESULTS SAVED: cluster1_imbalanced_WITH_SMOTE.csv")
print("="*80)
print("\n✅ COMPLETE: Cluster 1 used SMOTE balancing techniques")
print(f"✅ Dataset: {len(X)} samples (originally {y.sum()} claims, {(y==0).sum()} non-claims)")
print(f"✅ SMOTE balanced the minority class (No Claims) from 25 to {y.sum()} samples")
```
